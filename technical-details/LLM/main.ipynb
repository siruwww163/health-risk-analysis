### **Research Paper Search Process**  

To support the literature review and strengthen the theoretical foundation of this project, I conducted an extensive search for relevant research papers using the Google Scholar API through the `scholarly` library. This search focused on papers related to **machine learning in healthcare**, **health risk prediction**, and **chronic diseases management**.  

The primary objective was to gather high-impact papers that cover diverse aspects of the field, including supervised and unsupervised learning applications, feature engineering techniques, and public health implications. By querying terms such as "machine learning in healthcare," "health risk prediction," and "chronic disease management," I retrieved 50 relevant papers.  

Key bibliographic details, including **title**, **authors**, **publication venue**, **year of publication**, **abstract**, and **direct download links**, were extracted and organized into a structured CSV file. This systematic approach allowed efficient reference management, enabling comprehensive literature analysis to inform both the **methodology design** and the **interpretation of results**.  

The collected papers serve as a vital resource, providing insights into previous methodologies, existing challenges, and proposed solutions in the healthcare machine learning domain. This research-driven approach enhances the academic rigor of the project while ensuring its alignment with current advancements in the field.

```python
from scholarly import scholarly
import pandas as pd

# Define search query
query = "machine learning in healthcare health risk prediction chronic diseases"

# Search for papers
search_results = scholarly.search_pubs(query)

# Collect top 50 papers
papers = []
for i in range(50):
    try:
        paper = next(search_results)
        papers.append({
            'Title': paper['bib']['title'],
            'Authors': ', '.join(paper['bib']['author']),
            'Publication': paper['bib'].get('venue', 'N/A'),
            'Year': paper['bib'].get('year', 'N/A'),
            'Abstract': paper['bib'].get('abstract', 'N/A'),
            'Link': paper['eprint_url'] if 'eprint_url' in paper else "N/A"
        })
    except StopIteration:
        break

# Save to CSV
papers_df = pd.DataFrame(papers)
papers_df.to_csv('../../data/paper/research_paper', index=False)

print("Saved top 50 papers to research_papers.csv")
```
