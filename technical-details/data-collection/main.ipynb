## WHO Data Collection

The WHO API provides access to a comprehensive database of global health indicators, covering various aspects of public health, including chronic disease prevalence, risk factors, and treatment coverage. Since querying the API requires specific indicator codes, the first step involved fetching the full list of indicators using the [WHO Indicator API Endpoint](https://ghoapi.azureedge.net/api/Indicator). The data was saved as `who_indicators_list.csv` for further filtering. 

Relevant indicators were identified by searching for keywords such as "blood," "pressure," "glucose," "diabetes," and "hypertension." These terms align with the project’s focus on chronic disease analysis. After filtering, key indicators related to BMI, hypertension, and diabetes were selected based on their relevance to the research objectives. This streamlined dataset enabled targeted data collection using the WHO API.

The selected indicators were programmatically fetched using Python’s `requests` library and saved as CSV files. Each file represents a specific health indicator, structured with consistent columns for easy integration and analysis. These files include data on average BMI, diabetes prevalence, and hypertension control rates across countries, demographic groups, and years, forming the foundation for the project’s subsequent data analysis and modeling efforts.

```python
# WHO API main URL
base_url = "https://ghoapi.azureedge.net/api/Indicator"

# Request available indicators from WHO API
response = requests.get(base_url)

if response.status_code == 200:
    data = response.json()
    indicators = pd.DataFrame(data['value'])
    indicators.to_csv('../../data/raw-data/who_indicators_list.csv', index=False)
    print("WHO indicator list has been saved to who_indicators_list.csv")
else:
    print(f"Request failed, status code: {response.status_code}")
```

### Contents of `who_indicators_list`
The `who_indicators_list.csv` file contains metadata about all available health indicators in the WHO database. Each row represents a specific health indicator, with the following key columns:

- **IndicatorCode**: The unique identifier required to query the specific dataset.

- **IndicatorName**: A descriptive label that provides details about what the indicator measures.

- **Language**: The language in which the indicator name is written. 

This dataset offers a broad overview of available health statistics, spanning diverse topics such as environmental health, chronic diseases, financial protection, and global epidemiology. It serves as the foundation for selecting the specific indicators relevant to this project. By filtering the data for keywords associated with chronic diseases and allowed me to filter and identify the most relevant indicators for my project.

```python
# Load the CSV into a DataFrame
who_indicators_df = pd.read_csv('../../data/raw-data/who_indicators_list.csv')

# Display the first few rows to inspect the structure
who_indicators_df.head()
```

After obtaining the full list of indicators, I filtered them based on specific keywords such as **"blood"**, **"pressure"**, **"glucose"**, **"diabetes"**, and **"hypertension"**. These keywords align with the chronic disease focus of this project. The filtering process was automated using pandas library.

```python
# Filter the WHO indicators list for relevant keywords
keywords = ["blood", "pressure", "glucose", "diabetes", "hypertension"]
filtered_indicators = who_indicators_df[who_indicators_df['IndicatorName'].str.contains('|'.join(keywords), case=False, na=False)]

filtered_indicators
```

This step narrowed the list to a manageable subset of indicators directly related to the research objectives. The selected indicators include:

- BMI:

    - `NCD_BMI_MEAN`: Mean Body Mass Index across countries, reflecting obesity trends.

- Hypertension:
    - `NCD_HYP_PREVALENCE_A`: Prevalence of hypertension among adults.
    - `NCD_HYP_CONTROL_A`: Hypertension control rates.
    - `NCD_HYP_TREATMENT_A`: Hypertension treatment rates.

- Diabetes and Blood Glucose:
    - `NCD_DIABETES_PREVALENCE_CRUDE`: Crude diabetes prevalence.
    - `NCD_DIABETES_PREVALENCE_AGESTD`: Age-standardized diabetes prevalence.
    - `NCD_GLUC_01` and `NCD_GLUC_02`: Fasting blood glucose levels.

These indicators were chosen because they provide quantitative measures of chronic disease prevalence, risk factors, and management strategies, making them essential for understanding global health patterns.

```python
# WHO API request function
def fetch_who_data(indicator, output_file):
    url = f"https://ghoapi.azureedge.net/api/{indicator}"
    response = requests.get(url)

    if response.status_code == 200:
        data = response.json()
        # Extract relevant data fields
        df = pd.DataFrame(data['value'])
        df.to_csv(f'../../data/raw-data/{output_file}.csv', index=False)
        print(f"{output_file}.csv Data saved successfully！")
    else:
        print(f"Request failed, status code: {response.status_code}")

# Get blood pressure related data
fetch_who_data("NCD_BMI_MEAN", "who_bmi_data")
fetch_who_data("NCD_HYP_PREVALENCE_A", "who_hypertension_prevalence")
fetch_who_data("NCD_HYP_CONTROL_A", "who_hypertension_control")
fetch_who_data("NCD_HYP_TREATMENT_A", "who_hypertension_treatment")

# Get diabetes related data
fetch_who_data("NCD_BMI_MEAN", "who_bmi_data")
fetch_who_data("NCD_GLUC_01", "who_blood_glucose_age_std")
fetch_who_data("NCD_GLUC_02", "who_blood_glucose_crude")
fetch_who_data("NCD_DIABETES_PREVALENCE_CRUDE", "who_diabetes_prevalence_crude")
fetch_who_data("NCD_DIABETES_PREVALENCE_AGESTD", "who_diabetes_prevalence_age_std")
```

The raw data collected from the WHO API is stored in CSV format. Each file represents a specific health indicator (e.g., BMI, diabetes prevalence, hypertension control) and contains consistent column structures across all datasets. Data can be found from [WHO Data](https://drive.google.com/drive/folders/1puoOCSccRjK5w9MeDEmJUJ4dbnpPkFhC?usp=sharing).

Each CSV file corresponds to a specific health indicator :

1. **`who_bmi_data.csv`**:  
   - Mean Body Mass Index by country, year, and demographic group.

2. **`who_hypertension_prevalence.csv`**:  
   - Hypertension prevalence by geographic and demographic dimensions.

3. **`who_hypertension_control.csv`**:  
   - Effective hypertension treatment coverage, indicating the percentage of the population receiving adequate treatment.

4. **`who_hypertension_treatment.csv`**:  
   - Hypertension treatment coverage, reflecting the proportion of the population undergoing hypertension treatment.

5. **`who_diabetes_prevalence_crude.csv`**:  
   - Crude diabetes prevalence, providing an unadjusted rate of diabetes occurrence.

6. **`who_diabetes_prevalence_age_std.csv`**:  
   - Age-standardized diabetes prevalence, adjusted for age distribution differences.

7. **`who_blood_glucose_age_std.csv`**:  
   - Mean fasting blood glucose levels, age-standardized, representing the average glucose concentration among the population.

8. **`who_blood_glucose_crude.csv`**:  
   - Mean fasting blood glucose levels, crude estimates, reflecting the average glucose concentration without adjustments.

### Columns in the WHO Dataset

The WHO datasets contain a range of metadata, demographic details, numerical data, and additional context-specific information. Here’s a breakdown of the key columns found across the collected CSV files:  

#### **1. Metadata Columns:**  
- **`Id`**: Unique identifier for each observation (Integer).  
- **`IndicatorCode`**: Unique code representing the health indicator (String).  
- **`SpatialDimType` & `SpatialDim`**: Geographic classification, indicating country or region (String).  
- **`ParentLocationCode` & `ParentLocation`**: Higher-level geographic grouping for regional aggregation (String).  
- **`TimeDimType` & `TimeDim`**: Temporal classification indicating the year or time period of the data (String & Integer).  

#### **2. Demographic Dimensions:**  
- **`Dim1Type` & `Dim1`**: Primary demographic factor (e.g., gender, age group) (String).  
- **`Dim2Type` & `Dim2`**: Secondary demographic category, if applicable (String).  
- **`Dim3Type` & `Dim3`**: Tertiary demographic category, for further breakdown (String).  

#### **3. Data Source:**  
- **`DataSourceDimType` & `DataSourceDim`**: Information about the data source or collection method (String).  

#### **4. Numerical Data Columns:**  
- **`Value`**: Recorded health statistic, often provided as a descriptive range (String).  
- **`NumericValue`**: Central numerical value representing the indicator's exact measurement (Float).  
- **`Low` & `High`**: Confidence interval bounds indicating the reliability of the reported statistic (Float).  

#### **5. Additional Columns:**  
- **`Comments`**: Additional notes or remarks about the data entry (String).  
- **`Date`**: Date when the data was last updated (Datetime).  
- **`TimeDimensionBegin` & `TimeDimensionEnd`**: Start and end dates of the observation period, if applicable (Datetime).  

These columns provide essential context and structure, supporting a comprehensive understanding of global health indicators, demographic patterns, and trends crucial for the analysis of chronic disease prevalence.
